```python
#Import packages
import nltk
nltk.download('punkt')
from gensim.models.doc2vec import Doc2Vec, TaggedDocument
from nltk.tokenize import word_tokenize

# Example document (list of sentences)
doc = ["I love data science",
        "I love coding in python",
        "I love building NLP tool",
        "This is a good phone",
        "This is a good laptop"]

# Tokenization of each document
tokenized_doc = []
for d in doc:
    tokenized_doc.append(word_tokenize(d.lower()))
tokenized_doc
```

```python
# Convert tokenized document into gensim formated tagged data
tagged_data = [TaggedDocument(d, [i]) for i, d in enumerate(tokenized_doc)]
tagged_data
```

**dm = 0 -> PVDBOW**

**dm = 1 -> PVDM**

```python
## Train doc2vec model
model = Doc2Vec(tagged_data, vector_size=20, window=2, min_count=1, workers=4, dm=0, epochs = 100)
```

```python
## Print model vocabulary
model.wv.vocab
```

```python
# find most similar doc 
test_doc = word_tokenize("This is a good device".lower())
model.docvecs.most_similar(positive=[model.infer_vector(test_doc)],topn=5)
```
