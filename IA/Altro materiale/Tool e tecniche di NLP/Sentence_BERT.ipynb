# Generazione degli embedding


```python
import numpy as np

def cosine(u, v):
    return np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))
```

**Download modello SentenceBERT**

```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer('multi-qa-distilbert-dot-v1')
```

```python
sentence1 = "This is the final lesson of the AI course, I will see you all in January. In the meantime, I wish you all a Merry Christmas and a Happy New Year!"
sentence2 = "Today we finish our Artificial Intelligence course. Since I won't see you for over a month, I wish you all spend a happy Christmas holidays!"
sentence3 = "Christmas is coming, I can't wait to open my gifts and stay with my family. I will have to attend my exams in January, though."
vector = [sentence1, sentence2, sentence3]
```

```python
sentence_embeddings = model.encode(vector)
```

```python
print(sentence1 + '\n' + sentence2 + '\t\t->\t' + str(cosine(sentence_embeddings[0], sentence_embeddings[1]))+ '\n\n')
print(sentence1 + '\n' + sentence3 + '\t\t->\t' + str(cosine(sentence_embeddings[0], sentence_embeddings[2]))+ '\n\n')
print(sentence2 + '\n' + sentence3 + '\t\t->\t' + str(cosine(sentence_embeddings[1], sentence_embeddings[2]))+ '\n\n')
```
